## Problem
Build the detector a phishing-awareness tool would use. Given the raw text of a message, return
which of the four psychological pressure tactics it employs:

- `"urgency"`, manufactured time pressure
- `"authority"`, borrowed legitimacy (boss, IT, government, bank)
- `"scarcity"`, limited availability / fear of missing out
- `"intimidation"`, threats / fear

A message may use zero, one, or several tactics.

## Function contract
Implement `solve(text: str) -> list[str]` in the code editor (see the Reference section below).
Return the detected tactic labels as a **sorted, de-duplicated** list. Matching is
case-insensitive. Detection is by the indicator phrase sets defined in the Reference section below.

## Input
A single string, the message body (and/or subject) to analyze.

## Output
A sorted list of unique tactic labels from the set above (empty list if none).

## Examples
```
"Your account will be locked in 24 hours. Act now." -> ["urgency"]
"This is the IT department. Limited spots — last chance." -> ["authority", "scarcity"]
"Pay the penalty or we will take legal action immediately" -> ["intimidation", "urgency"]
"Lunch is in the kitchen, help yourself." -> []
```

### Input (`text`)

### Output (`result`)
A **sorted, de-duplicated** `list[str]` drawn from
`{"authority", "intimidation", "scarcity", "urgency"}`. Empty list if no tactic is present.

### Indicator phrase sets (case-insensitive substring match)
- **urgency:** `urgent`, `act now`, `immediately`, `right now`, `asap`, `expires`, `within 24 hours`,
  `24 hours`, `as soon as possible`, `don't delay`, `do not delay`
- **authority:** `it department`, `ceo`, `manager`, `administrator`, `irs`, `bank`, `police`,
  `official`, `help desk`, `human resources`, `hr department`
- **scarcity:** `limited`, `only`, `last chance`, `running out`, `few left`, `exclusive`,
  `while supplies last`, `act fast`
- **intimidation:** `legal action`, `account suspended`, `account will be locked`, `penalty`,
  `arrested`, `fired`, `terminated`, `lawsuit`, `consequences`, `reported`

### Guarantees / rules
- Pure function, deterministic, no I/O, no network. Standard library only.
- Matching is case-insensitive. Output always sorted ascending and de-duplicated.

### Acceptance
Your code passes when every test below shows a checkmark (some are hidden until you run them).


````python
# Reference solution
"""Reference solution for 1.1.a. MUST pass 100% of the tests.
Also the default 'blackbox' implementation students can swap out (see the hidden grading tests)."""
from typing import List

INDICATORS = {
    "urgency": [
        "urgent", "act now", "immediately", "right now", "asap", "expires",
        "within 24 hours", "24 hours", "as soon as possible", "don't delay", "do not delay",
    ],
    "authority": [
        "it department", "ceo", "manager", "administrator", "irs", "bank", "police",
        "official", "help desk", "human resources", "hr department",
    ],
    "scarcity": [
        "limited", "only", "last chance", "running out", "few left", "exclusive",
        "while supplies last", "act fast",
    ],
    "intimidation": [
        "legal action", "account suspended", "account will be locked", "penalty", "arrested",
        "fired", "terminated", "lawsuit", "consequences", "reported",
    ],
}


def solve(text: str) -> List[str]:
    """Return the sorted, de-duplicated psychological tactics present in `text`."""
    low = (text or "").lower()
    found = {tactic for tactic, phrases in INDICATORS.items()
             if any(p in low for p in phrases)}
    return sorted(found)
````
