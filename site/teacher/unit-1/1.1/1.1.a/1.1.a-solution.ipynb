## 3. Build it

This is the model solution.

- [ ] `solve` handles the worked examples
- [ ] `solve` handles your edge cases

````python
from typing import List

INDICATORS = {
    "urgency": [
        "urgent", "act now", "immediately", "right now", "asap", "expires",
        "within 24 hours", "24 hours", "as soon as possible", "don't delay", "do not delay",
    ],
    "authority": [
        "it department", "ceo", "manager", "administrator", "irs", "bank", "police",
        "official", "help desk", "human resources", "hr department",
    ],
    "scarcity": [
        "limited", "only", "last chance", "running out", "few left", "exclusive",
        "while supplies last", "act fast",
    ],
    "intimidation": [
        "legal action", "account suspended", "account will be locked", "penalty", "arrested",
        "fired", "terminated", "lawsuit", "consequences", "reported",
    ],
}


def solve(text: str) -> List[str]:
    low = (text or "").lower()
    found = {tactic for tactic, phrases in INDICATORS.items()
             if any(p in low for p in phrases)}
    return sorted(found)
````

## 4. Test it

The tests are the acceptance check, like a change review before a fix goes live.

- [ ] Run the tests
- [ ] Every test says PASS (if not, read the message, go back to step 3)

````python
import __main__ as impl

def test_urgency_and_intimidation():
    assert impl.solve("Your account will be locked in 24 hours. Act now.") == ["intimidation", "urgency"]


def test_authority_and_scarcity():
    assert impl.solve("This is the IT department. Limited spots - last chance.") == ["authority", "scarcity"]


def test_intimidation_and_urgency():
    assert impl.solve("Pay the penalty or we will take legal action immediately") == ["intimidation", "urgency"]


def test_none():
    assert impl.solve("Lunch is in the kitchen, help yourself.") == []


def test_case_insensitive():
    assert impl.solve("URGENT: the CEO needs this ASAP") == ["authority", "urgency"]


def test_dedup_sorted():
    out = impl.solve("urgent urgent act now - exclusive offer, only a few left")
    assert out == ["scarcity", "urgency"]
    assert out == sorted(set(out))


def test_empty_string():
    assert impl.solve("") == []


# ---- run every test above ------------------------------------------------
_results = []
for _name, _fn in list(globals().items()):
    if _name.startswith("test_") and callable(_fn):
        try:
            _fn()
            _results.append((_name, None))
        except Exception as _e:
            _results.append((_name, _e))
for _name, _e in _results:
    print(("PASS  " if _e is None else "FAIL  ") + _name + ("" if _e is None else f": {_e!r}"))
print(f"{sum(e is None for _, e in _results)}/{len(_results)} tests passed")
````
