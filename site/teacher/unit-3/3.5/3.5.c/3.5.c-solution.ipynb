## 1. Brief

**The situation.** The NIST SP 800-61 incident response lifecycle has six phases, in this canonical order:

```
1. Preparation
2. Identification
3. Containment
4. Eradication
5. Recovery
6. Lessons Learned
```

You will implement a function that takes a list of response-action records, each tagged with the
phase it belongs to, but given in shuffled order, and returns the actions sorted into correct
lifecycle order. Actions within the same phase keep their original relative order (a stable sort).

An **action record** is a dict with:
- `"action"`: string description of the action
- `"phase"`: string, one of the six canonical phase names exactly as spelled above

**Lesson objective.** Name the six phases of the NIST incident-response lifecycle in order, classify a given response action into its correct phase, and explain why performing actions out of order (for example, eradicating before containing) can make an incident worse.

## 3. Build it

This is the model solution.

- [ ] `solve` handles the worked examples
- [ ] `solve` handles your edge cases

````python
from typing import Dict, List

PHASE_ORDER = [
    "Preparation",
    "Identification",
    "Containment",
    "Eradication",
    "Recovery",
    "Lessons Learned",
]
_PHASE_RANK = {name: i for i, name in enumerate(PHASE_ORDER)}


def solve(actions: List[Dict]) -> List[str]:
    for entry in actions:
        if entry["phase"] not in _PHASE_RANK:
            raise ValueError(f"Unknown IR phase: {entry['phase']!r}")

    ordered = sorted(actions, key=lambda entry: _PHASE_RANK[entry["phase"]])
    return [entry["action"] for entry in ordered]
````

## 4. Test it

The tests are the acceptance check, like a change review before a fix goes live.

- [ ] Run the tests
- [ ] Every test says PASS (if not, read the message, go back to step 3)

````python
import __main__ as impl

import pytest


def test_basic_reordering():
    actions = [
        {"action": "Rebuild the server from a clean backup", "phase": "Recovery"},
        {"action": "Write the incident response plan", "phase": "Preparation"},
        {"action": "Isolate the infected host from the network", "phase": "Containment"},
        {"action": "Confirm the IDS alert is a real compromise", "phase": "Identification"},
    ]
    assert impl.solve(actions) == [
        "Write the incident response plan",
        "Confirm the IDS alert is a real compromise",
        "Isolate the infected host from the network",
        "Rebuild the server from a clean backup",
    ]


def test_all_six_phases():
    actions = [
        {"action": "Document a timeline and update the IR plan", "phase": "Lessons Learned"},
        {"action": "Remove the backdoor account", "phase": "Eradication"},
        {"action": "Reconnect the rebuilt host and monitor", "phase": "Recovery"},
        {"action": "Run a tabletop exercise", "phase": "Preparation"},
        {"action": "Block the malicious IP at the firewall", "phase": "Containment"},
        {"action": "Triage the alert queue", "phase": "Identification"},
    ]
    result = impl.solve(actions)
    assert result == [
        "Run a tabletop exercise",
        "Triage the alert queue",
        "Block the malicious IP at the firewall",
        "Remove the backdoor account",
        "Reconnect the rebuilt host and monitor",
        "Document a timeline and update the IR plan",
    ]


def test_stable_within_same_phase():
    actions = [
        {"action": "Contain host A", "phase": "Containment"},
        {"action": "Contain host B", "phase": "Containment"},
        {"action": "Prep step 1", "phase": "Preparation"},
    ]
    assert impl.solve(actions) == ["Prep step 1", "Contain host A", "Contain host B"]


def test_already_in_order_unchanged():
    actions = [
        {"action": "p", "phase": "Preparation"},
        {"action": "i", "phase": "Identification"},
        {"action": "c", "phase": "Containment"},
    ]
    assert impl.solve(actions) == ["p", "i", "c"]


def test_invalid_phase_raises_value_error():
    actions = [{"action": "Bogus phase", "phase": "Investigation"}]
    with pytest.raises(ValueError):
        impl.solve(actions)


def test_empty_list():
    assert impl.solve([]) == []


def test_single_action():
    assert impl.solve([{"action": "only one", "phase": "Recovery"}]) == ["only one"]


# ---- run every test above ------------------------------------------------
_results = []
for _name, _fn in list(globals().items()):
    if _name.startswith("test_") and callable(_fn):
        try:
            _fn()
            _results.append((_name, None))
        except Exception as _e:
            _results.append((_name, _e))
for _name, _e in _results:
    print(("PASS  " if _e is None else "FAIL  ") + _name + ("" if _e is None else f": {_e!r}"))
print(f"{sum(e is None for _, e in _results)}/{len(_results)} tests passed")
````
