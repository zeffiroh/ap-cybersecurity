## Problem
You are building a control matrix tool for Crestfield Analytics' security program. Given a list of
implemented security scenarios, classify each one by control family: `"managerial"`,
`"technical"`, or `"physical"`, based on how the control is actually implemented (its
`"mechanism"`), not on what risk it happens to address.

## Function contract
Implement `classify_controls(scenarios)` in the code editor (see the Reference section below).

## Input
- `scenarios`: `list[dict]`, each with:
  - `"id"`: `str`, a scenario identifier
  - `"description"`: `str`, flavor text describing the scenario (not used for classification)
  - `"mechanism"`: `str`, one of `"policy_document"`, `"training_program"`,
    `"access_review_process"`, `"software_or_hardware_enforcement"`, `"physical_barrier"`

## Output
A `list[str]` of classifications, **one per input scenario, in the same order as the input**. Each
entry is one of `"managerial"`, `"technical"`, `"physical"`.

## Examples
```python
scenarios = [
    {"id": "s1", "description": "New hires sign a document listing allowed and prohibited use of company laptops.",
     "mechanism": "policy_document"},
    {"id": "s2", "description": "The office network blocks known-malicious IP ranges automatically.",
     "mechanism": "software_or_hardware_enforcement"},
    {"id": "s3", "description": "The server room door requires a badge to open.",
     "mechanism": "physical_barrier"},
]
classify_controls(scenarios)  ->  ["managerial", "technical", "physical"]
```

## Constraints
- Deterministic, offline, standard library only.
- Classification depends only on `"mechanism"`, never on `"description"` text.
- Mapping: `"policy_document"`, `"training_program"`, and `"access_review_process"` are all
  `"managerial"`; `"software_or_hardware_enforcement"` is `"technical"`; `"physical_barrier"` is
  `"physical"`.
- Must pass all cases in the tests (visible + hidden).

## Reference

### Entry point
```python
def classify_controls(scenarios: list[dict]) -> list[str]
```

### Input (`scenarios`)
`list[dict]` where each dict has:
- `"id"`: `str`
- `"description"`: `str`, flavor text, not used for classification logic
- `"mechanism"`: `str`, one of `"policy_document"`, `"training_program"`,
  `"access_review_process"`, `"software_or_hardware_enforcement"`, `"physical_barrier"`

### Output (`result`)
A `list[str]` the same length as `scenarios`, in the **same order as the input**, where each
element is one of `"managerial"`, `"technical"`, `"physical"`.

### Classification rule
- `mechanism` in `{"policy_document", "training_program", "access_review_process"}` -> `"managerial"`
- `mechanism == "software_or_hardware_enforcement"` -> `"technical"`
- `mechanism == "physical_barrier"` -> `"physical"`

### Guarantees / rules
- Pure function, deterministic, no I/O, no network. Standard library only.
- `"mechanism"` always contains one of the five listed values in test data (no invalid values).
- Output order matches input order exactly.

### Acceptance
Your code passes when every test below shows a checkmark (some are hidden until you run them).


````python
# Reference solution
"""Reference solution for 3.2.b. MUST pass 100% of the tests.
Also the default 'blackbox' implementation students can swap out (see the hidden grading tests)."""
from typing import Dict, List

_MANAGERIAL = {"policy_document", "training_program", "access_review_process"}
_TECHNICAL = {"software_or_hardware_enforcement"}
_PHYSICAL = {"physical_barrier"}


def classify_controls(scenarios: List[Dict]) -> List[str]:
    """Classify each scenario's control family by its 'mechanism' field.

    Rules (see the Reference section below):
      - policy_document / training_program / access_review_process -> managerial
      - software_or_hardware_enforcement -> technical
      - physical_barrier -> physical
    """
    result = []
    for s in scenarios:
        mech = s["mechanism"]
        if mech in _MANAGERIAL:
            result.append("managerial")
        elif mech in _TECHNICAL:
            result.append("technical")
        elif mech in _PHYSICAL:
            result.append("physical")
        else:
            raise ValueError(f"unknown mechanism: {mech}")
    return result
````
