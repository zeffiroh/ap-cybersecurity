## 1. Brief

**The situation.** In physical security auditing, we look at the combination of active controls to identify vulnerabilities or gaps in a layered defense. You are tasked with implementing a function that takes a description of a facility's physical security inventory and returns all identified security gaps based on audit rules.

Implement `solve(data: dict[str, bool]) -> list[str]` in the code editor.

**Lesson objective.** Identify the purpose of each physical access control layer, locks, badge readers, biometrics, and mantraps, explain how they combine into a defense-in-depth strategy, and pinpoint gaps when one or more controls are absent.

## 3. Build it

This is the model solution.

- [ ] `solve` handles the worked examples
- [ ] `solve` handles your edge cases

````python
def solve(data: dict[str, bool]) -> list[str]:
    gaps = []
    
    if not data.get("keyed_lock", False) and not data.get("badge_reader", False):
        gaps.append("no_key_control")
        
    if not data.get("badge_reader", False) and not data.get("biometric_reader", False):
        gaps.append("no_electronic_access")
        
    if not data.get("mantrap", False):
        gaps.append("no_tailgating_prevention")
        
    if not data.get("cctv", False):
        gaps.append("no_visual_monitoring")
        
    if not data.get("security_guard", False) and not data.get("biometric_reader", False):
        gaps.append("no_identity_verification")
        
    return sorted(gaps)
````
