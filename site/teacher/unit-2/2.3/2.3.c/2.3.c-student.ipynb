## 1. Brief

**The situation.** You are auditing the physical security posture of fictional facilities for Novara Data Center. Each
facility supplies a list of security **layers** currently in place. Each layer belongs to exactly
one **control category**: `"perimeter"`, `"personnel"`, `"electronic"`, `"barrier"`, or
`"monitoring"` (see atom.md, these are the five categories a sound defense-in-depth plan should
draw from). A facility's **depth score** is the number of *distinct* categories represented among
its layers, not the total number of layers. Two guards on patrol (both `"personnel"`) only count
once toward depth, because they do not close a new gap the way a different category would.

Implement `solve(layers)` that audits one facility's layer list and returns a report:

- `depth_score`: count of distinct valid categories present.
- `categories_present`: sorted list of the distinct valid category names present.
- `redundant_categories`: dict mapping each category that has *more than one* layer to the count
  of *extra* layers beyond the first (e.g. 3 personnel layers -> `{"personnel": 2}`). Categories
  with only one layer are omitted.
- `recommendation`: one of `"STRONG"`, `"ADEQUATE"`, `"WEAK"`, based on `depth_score`:
  - `depth_score >= 5` -> `"STRONG"`
  - `3 <= depth_score <= 4` -> `"ADEQUATE"`
  - `depth_score <= 2` -> `"WEAK"`

Unknown/invalid category strings (anything other than the five listed above) must be ignored
entirely, they do not count toward `depth_score`, do not appear in `categories_present`, and do
not count toward `redundant_categories`.

**Lesson objective.** Explain how security guards function as an active personnel control and how combining multiple control categories creates a defense-in-depth posture that raises the cost of physical attacks on a facility.

### What comes in
- `layers`: `list[dict]`, each dict has:
  - `"name"`: `str`, free-text description, e.g. `"lobby reception guard"`
  - `"category"`: `str`, one of `"perimeter"`, `"personnel"`, `"electronic"`, `"barrier"`,
    `"monitoring"` (or an invalid string, which must be ignored per above)

### What must come out
A `dict` with keys `"depth_score"` (`int`), `"categories_present"` (`list[str]`, sorted ascending),
`"redundant_categories"` (`dict[str, int]`), and `"recommendation"` (`str`).

````python
import __main__ as impl

def _layer(name, category):
    return {"name": name, "category": category}


def test_three_distinct_categories_adequate():
    layers = [
        _layer("chain-link fence + lighting", "perimeter"),
        _layer("badge reader on server room", "electronic"),
        _layer("lobby reception guard", "personnel"),
    ]
    result = impl.solve(layers)
    assert result == {
        "depth_score": 3,
        "categories_present": ["electronic", "perimeter", "personnel"],
        "redundant_categories": {},
        "recommendation": "ADEQUATE",
    }


def test_redundant_personnel_weak():
    layers = [
        _layer("lobby guard", "personnel"),
        _layer("roving night patrol", "personnel"),
        _layer("badge reader", "electronic"),
    ]
    result = impl.solve(layers)
    assert result == {
        "depth_score": 2,
        "categories_present": ["electronic", "personnel"],
        "redundant_categories": {"personnel": 1},
        "recommendation": "WEAK",
    }


def test_all_five_categories_strong():
    layers = [
        _layer("perimeter fence", "perimeter"),
        _layer("lobby guard", "personnel"),
        _layer("badge reader", "electronic"),
        _layer("mantrap", "barrier"),
        _layer("cctv feed", "monitoring"),
    ]
    result = impl.solve(layers)
    assert result["depth_score"] == 5
    assert result["categories_present"] == [
        "barrier", "electronic", "monitoring", "perimeter", "personnel",
    ]
    assert result["redundant_categories"] == {}
    assert result["recommendation"] == "STRONG"


def test_empty_layers():
    result = impl.solve([])
    assert result == {
        "depth_score": 0,
        "categories_present": [],
        "redundant_categories": {},
        "recommendation": "WEAK",
    }


def test_single_category_weak():
    layers = [_layer("badge reader", "electronic")]
    result = impl.solve(layers)
    assert result["depth_score"] == 1
    assert result["categories_present"] == ["electronic"]
    assert result["redundant_categories"] == {}
    assert result["recommendation"] == "WEAK"


def test_invalid_category_ignored():
    layers = [
        _layer("badge reader", "electronic"),
        _layer("lobby guard", "personnel"),
        _layer("guard dog", "canine"),
    ]
    result = impl.solve(layers)
    assert result["depth_score"] == 2
    assert result["categories_present"] == ["electronic", "personnel"]
    assert "canine" not in result["categories_present"]
    assert result["recommendation"] == "WEAK"


def test_four_categories_adequate_boundary():
    layers = [
        _layer("fence", "perimeter"),
        _layer("guard", "personnel"),
        _layer("badge reader", "electronic"),
        _layer("mantrap", "barrier"),
    ]
    result = impl.solve(layers)
    assert result["depth_score"] == 4
    assert result["recommendation"] == "ADEQUATE"


def test_two_categories_multiple_redundancies():
    layers = [
        _layer("guard A", "personnel"),
        _layer("guard B", "personnel"),
        _layer("guard C", "personnel"),
        _layer("cctv 1", "monitoring"),
        _layer("cctv 2", "monitoring"),
    ]
    result = impl.solve(layers)
    assert result["depth_score"] == 2
    assert result["categories_present"] == ["monitoring", "personnel"]
    assert result["redundant_categories"] == {"personnel": 2, "monitoring": 1}
    assert result["recommendation"] == "WEAK"


def test_only_invalid_categories_is_weak_empty():
    layers = [_layer("unicorn detector", "magic"), _layer("weather vane", "meteorology")]
    result = impl.solve(layers)
    assert result == {
        "depth_score": 0,
        "categories_present": [],
        "redundant_categories": {},
        "recommendation": "WEAK",
    }


def test_categories_present_sorted_regardless_of_input_order():
    layers = [
        _layer("cctv", "monitoring"),
        _layer("fence", "perimeter"),
        _layer("mantrap", "barrier"),
    ]
    result = impl.solve(layers)
    assert result["categories_present"] == ["barrier", "monitoring", "perimeter"]


# ---- run every test above ------------------------------------------------
_results = []
for _name, _fn in list(globals().items()):
    if _name.startswith("test_") and callable(_fn):
        try:
            _fn()
            _results.append((_name, None))
        except Exception as _e:
            _results.append((_name, _e))
for _name, _e in _results:
    print(("PASS  " if _e is None else "FAIL  ") + _name + ("" if _e is None else f": {_e!r}"))
print(f"{sum(e is None for _, e in _results)}/{len(_results)} tests passed")
````
