## Problem

Categories:
- `"virus"`, attaches to host files; spreads when files are shared or opened
- `"worm"`, self-propagates across the network without user action; autonomous scanning
- `"trojan"`, disguised as legitimate software; creates backdoor / exfiltrates data
- `"ransomware"`, encrypts files; demands payment for decryption key

A report may match exactly one category. If no category matches, return `"unknown"`.
