# Network lab 2: Two networks and a router (Network Starter)

Join two office networks with a router, set every computer's gateway, and trace a packet's path across it.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `netbuilder.json`, `README.txt`.

## 1. Cable both networks

The shop (till-1, till-2 and switch **sw-shop**), the back office (accounts, manager and **sw-back**) and one **router** are already on the canvas. In **Design** mode, choose **Cable** and join each computer to its own switch. Then cable the router to **sw-shop first** and to **sw-back** second: the router's first interface, G0/0, then faces the shop and G0/1 faces the back office.

- [ ] Each computer cabled to its own switch
- [ ] Router cabled to sw-shop, then to sw-back
- [ ] The first goal is ticked

> Hint: Each router interface belongs to one network, so each needs its own address. That's why a router joins networks and a switch doesn't.

## 2. Set the router and the gateways

Click the router: its panel lists its interfaces. Give **G0/0** `192.168.1.1/24` and **G0/1** `192.168.2.1/24`. Now click each computer and type its address from README.txt and its **Default gateway**: shop computers use 192.168.1.1, back-office computers use 192.168.2.1.

The gateway is where a computer sends anything that isn't in its own network.

- [ ] Router interfaces set
- [ ] Shop gateways 192.168.1.1
- [ ] Back-office gateways 192.168.2.1

> Hint: A gateway must be inside the computer's own network. 192.168.2.1 is useless as a gateway for till-1.

## 3. Ping across, then break a gateway

Press **Simulation**, click till-1 and open **Command line**. Ping accounts at 192.168.2.21: the replies say **TTL=127**. Now open the **Data exchange** tab, choose **accounts** in the list, and click an ICMP echo request. Its **Time to live** is 127, one less than the 128 till-1 sent, because the router counted the hop. Its source MAC isn't till-1's either: the router put its own on.

Then press **Design**, clear manager's **Default gateway**, and press **Simulation**. From manager, ping accounts (192.168.2.21, same network): it answers. Ping till-1 (192.168.1.11): **General failure**. Put the gateway back.

- [ ] till-1 to accounts works
- [ ] TTL 127 read in accounts' data exchange
- [ ] Without a gateway, manager only reaches its own network

> Hint: A missing gateway is one of the most common faults on real help desks: "I can print but I can't reach the internet".
