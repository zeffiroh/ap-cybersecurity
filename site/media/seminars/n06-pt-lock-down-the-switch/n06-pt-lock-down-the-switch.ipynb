# Network lab 6: Lock down the switch (Network Simulator)

Shut unused ports, allow one known device per staff port with port security, protect the console and enable modes, and watch an unknown laptop get blocked.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `netbuilder.json`, `README.txt`.

## 1. Park and shut the unused ports

A live, empty wall port is an open door: anyone can plug in. Double-click **sw1** for its CLI:

```
enable
configure terminal
vlan 999
 name PARKING
interface range fa0/21 - 24
 switchport mode access
 switchport access vlan 999
 shutdown
end
show interfaces status
```

- [ ] VLAN 999 made
- [ ] Ports 21-24 in VLAN 999 and shut down
- [ ] show interfaces status shows them disabled

> Hint: Shut ports show as "disabled" in `show interfaces status`.

## 2. One known device per staff port

```
configure terminal
interface range fa0/1 - 12
 switchport port-security
 switchport port-security maximum 1
 switchport port-security mac-address sticky
 switchport port-security violation shutdown
end
```

From staff-1's Command Prompt, ping staff-2 (192.168.10.12), so both staff ports hear from their computers, then run `show port-security` on the switch. (On a real switch, a port that has never heard from its computer learns whatever plugs in first.)

- [ ] Port security on ports 1-12
- [ ] Fa0/1 and Fa0/2 each show 1 current address
- [ ] show port-security checked

> Hint: Sticky learns the first MAC address seen on the port and keeps it in the configuration.

## 3. Plug in an unknown laptop

Click the cable between staff-2 and the switch and press **Unplug it**. Add a **Laptop** from the bar, give it 192.168.10.50 in its panel, and cable it to **Fa0/2** with **Copper Straight-Through**. Ping staff-1 from the laptop. Look at the link light, then run `show interfaces status` and `show port-security` on the switch.

The port has shut itself down. To bring it back after a check, unplug the laptop, plug staff-2 back into Fa0/2, and on that port run `shutdown` then `no shutdown` (on a real switch the port stays err-disabled until you do).

- [ ] Unknown laptop blocked
- [ ] Port state read in show interfaces status
- [ ] staff-2 back on Fa0/2 and connected

> Hint: The link light turns red and the status says err-disabled.

## 4. Protect the management access

Anyone with a console cable could change all of this. Add passwords:

```
configure terminal
enable secret <choose one>
line console 0
 password <choose another>
 login
end
copy running-config startup-config
show running-config
```

Read the top of `show running-config`: the enable secret is stored as a hash (`enable secret 5 ...`), never as the word you typed.

- [ ] enable secret set
- [ ] Console password with login
- [ ] Configuration saved ([OK])

> Hint: `enable secret` is stored as a hash; the old `enable password` isn't. Always use secret.
