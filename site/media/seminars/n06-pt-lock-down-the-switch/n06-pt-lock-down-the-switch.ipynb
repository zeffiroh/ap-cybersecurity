# Network lab 6: Lock down the switch (Packet Tracer)

Shut unused ports, allow one known device per staff port with port security, protect the console and enable modes, and watch an unknown laptop get blocked.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `README.txt`.

## 1. Park and shut the unused ports

A live, empty wall port is an open door: anyone can plug in. On the switch:

```
enable
configure terminal
vlan 999
 name PARKING
interface range fa0/21 - 24
 switchport mode access
 switchport access vlan 999
 shutdown
end
show interfaces status
```

- [ ] VLAN 999 made
- [ ] Ports 21-24 in VLAN 999 and shut down
- [ ] show interfaces status shows them disabled

> Hint: Shut ports show as "disabled" in `show interfaces status`.

## 2. One known device per staff port

```
configure terminal
interface range fa0/1 - 12
 switchport port-security
 switchport port-security maximum 1
 switchport port-security mac-address sticky
 switchport port-security violation shutdown
end
```

From staff-1, ping staff-2 (192.168.10.12), so the switch learns both computers' MAC addresses, then run `show port-security interface fa0/1`. (If staff-2 never sends anything, Fa0/2 learns whatever plugs in first, and step 3 won't be blocked.)

- [ ] Port security on ports 1-12
- [ ] staff-1's MAC learned (sticky)
- [ ] show port-security checked

> Hint: Sticky learns the first MAC address seen on the port and keeps it in the configuration.

?> How many MAC addresses may each staff port learn? :: sha256:6b86b273ff34fce19d6b804eff5a3f5747ada4eaa22f1d49c01e52ddb7875b4b

## 3. Plug in an unknown laptop

Unplug staff-2 from Fa0/2 and plug a new **Laptop** into Fa0/2 instead. Give it 192.168.10.50 and ping staff-1. Look at the link light, then run `show interfaces status` and `show port-security interface fa0/2`.

The port has shut itself down. To bring it back after a check, plug staff-2 back in and run `shutdown` then `no shutdown` on the port.

- [ ] Unknown laptop blocked
- [ ] Port state read in show interfaces status
- [ ] Port brought back with shutdown / no shutdown

> Hint: In Packet Tracer the light turns red and the status says err-disabled.

## 4. Protect the management access

Anyone with a console cable could change all of this. Add passwords:

```
configure terminal
enable secret <choose one>
line console 0
 password <choose another>
 login
end
copy running-config startup-config
```

Test it: open a console from a PC (**Console** cable to the switch's console port, then **Desktop → Terminal**).

- [ ] enable secret set
- [ ] Console password with login
- [ ] Configuration saved

> Hint: `enable secret` is stored as a hash; the old `enable password` isn't. Always use secret.
