# Network lab 4: VLANs for staff and guests (Network Simulator)

Split one switch into a staff VLAN and a guest VLAN with real switch commands, and prove that guests can't reach staff computers.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `netbuilder.json`, `README.txt`.

## 1. Open Network Simulator and cable the office

Labs 4 to 7 use **Network Simulator**, which speaks the real commands of Cisco switches and routers. Nothing to install: in the Hacker Room open **Start → Workplace → Network lab** and pick lab 4. The **2960** switch and four **PCs** are already on the canvas.

In the **Cable** list choose **Copper Straight-Through**, so you can pick the ports. Press **Connect**, click a PC, then **sw1**, choose its port from README.txt and press **Connect**. Click each PC and type its **IP address** and **Subnet mask** 255.255.255.0 in the panel on the right.

## 2. Make the VLANs

Double-click **sw1** (or open the **CLI and Desktop** tab and pick sw1), and type:

```
enable
configure terminal
vlan 10
 name STAFF
vlan 20
 name GUEST
interface range fa0/1 - 12
 switchport mode access
 switchport access vlan 10
interface range fa0/13 - 20
 switchport mode access
 switchport access vlan 20
end
show vlan brief
```

- [ ] Both VLANs exist with names
- [ ] Ports 1-12 in VLAN 10, 13-20 in VLAN 20
- [ ] show vlan brief checked

> Hint: `interface range` sets many ports at once. Unused ports 21-24 stay in VLAN 1: you lock those down in lab 6.

## 3. Prove the separation

Double-click **staff-1** for its **Command Prompt**. Ping staff-2 (192.168.10.12): it answers. Ping guest-1 (192.168.20.31): it fails.

Now the real test: click guest-1 and change its address to 192.168.10.99, as if a visitor guessed the staff range. From guest-1's Command Prompt, ping staff-1. It still fails, because the switch never forwards VLAN 20 frames to VLAN 10 ports. Put the address back.

- [ ] Staff to staff works
- [ ] Staff to guest fails
- [ ] A guest using a staff address still can't reach staff

> Hint: VLANs separate at layer 2. Choosing a different IP address doesn't move you into another VLAN: the port decides.
