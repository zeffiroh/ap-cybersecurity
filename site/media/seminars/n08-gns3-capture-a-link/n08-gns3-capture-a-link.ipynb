# Network lab 8: Capture a link (Network Emulator)

Build a tiny network in Network Emulator with its virtual PCs and switch, start a capture on a link, and read ARP and ICMP packets layer by layer.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `netbuilder.json`, `README.txt`.

## 1. Open Network Emulator, start and link three nodes

Lab 8 uses **Network Emulator**: nodes that start and stop like real virtual machines, and a capture on any link. Nothing to install: in the Hacker Room open **Start → Workplace → Network lab** and pick lab 8. Two **VPCS** computers and an **Ethernet switch** are on the canvas, stopped (red).

Press **Start all**. Then press **Add a link**, click PC1, then SW1, pick the ports and press **Connect**. Do the same for PC2. Both ends of each link turn green.

- [ ] Network Emulator open with PC1, PC2 and SW1
- [ ] All three nodes started (green in the Topology summary)
- [ ] PC1 and PC2 linked to SW1

> Hint: A stopped node has no console and its link ends stay red. Right-click a node to start or stop just that one.

## 2. Give the computers addresses

Double-click PC1 (or right-click it → **Console**), and type `ip 10.0.0.1/24 10.0.0.254`. On PC2: `ip 10.0.0.2/24 10.0.0.254`. Check each with `show ip`. Each console gets its own tab in **Consoles**.

- [ ] PC1 and PC2 addressed
- [ ] show ip checked on both, and PC2's MAC noted

> Hint: The VPCS `ip` command takes the address with its prefix length, then the gateway.

## 3. Capture and read the packets

Right-click the link between PC1 and SW1 → **Start capture**. In PC1's console, type `ping 10.0.0.2`. Open the **Capture** tab, type `arp` in the **Display filter** box and press Enter: an **ARP** request (who has 10.0.0.2?) and its reply. Click the request and read its Ethernet destination. Then filter `icmp`: five **Echo (ping) request** and **reply** pairs. Click a request and read **Internet Control Message Protocol** in the details to see its type; click a reply and read its **Time to live**.

- [ ] Capture running on the PC1 link
- [ ] arp filter: the request and reply found
- [ ] icmp filter: echo request and reply pairs found
- [ ] ICMP type and TTL read in the details

> Hint: Echo request is ICMP type 8, echo reply is type 0.
