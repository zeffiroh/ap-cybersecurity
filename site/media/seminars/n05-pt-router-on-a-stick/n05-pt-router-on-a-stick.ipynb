# Network lab 5: Router on a stick (Network Simulator)

Carry both VLANs to a router over one trunk, give each VLAN a gateway on a subinterface, and route between them.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `netbuilder.json`, `README.txt`.

## 1. A trunk to the router

Lab 4's office is already built, and a **1941** router is waiting below the switch. With **Copper Straight-Through** chosen, press **Connect**, click sw1, then the router, and pick switch **Gi0/1** and router **G0/0**. On the switch CLI:

```
enable
configure terminal
interface gi0/1
 switchport mode trunk
end
show interfaces trunk
```

A trunk carries many VLANs on one cable by tagging each frame with its VLAN number (802.1Q).

- [ ] Router cabled to Gi0/1
- [ ] Gi0/1 is a trunk
- [ ] show interfaces trunk shows Gi0/1 trunking

> Hint: An access port carries one VLAN untagged. A trunk carries several, tagged.

## 2. A gateway per VLAN

Double-click the router for its CLI:

```
enable
configure terminal
interface g0/0
 no shutdown
interface g0/0.10
 encapsulation dot1Q 10
 ip address 192.168.10.1 255.255.255.0
interface g0/0.20
 encapsulation dot1Q 20
 ip address 192.168.20.1 255.255.255.0
end
show ip interface brief
```

Then click each PC and set its **Default gateway**: staff 192.168.10.1, guests 192.168.20.1.

- [ ] Main interface up
- [ ] Two subinterfaces with dot1Q tags and addresses
- [ ] Gateways set on all four PCs

> Hint: Router interfaces start shut down. The subinterfaces only work once the main interface has `no shutdown`.

## 3. Test, and notice the new risk

From guest-1's Command Prompt, ping staff-1 (192.168.10.11). It works now, because the router joins the VLANs. Good for printing; bad for security: guests can reach staff computers again.

Write down which traffic guests should be allowed (the internet) and which they shouldn't (staff computers and the file server). Lab 7 enforces that with an access list.

- [ ] Guest to staff ping works through the router

> Hint: Routing between VLANs is all or nothing until you add rules. That's why segmentation always comes with an access list or a firewall.
