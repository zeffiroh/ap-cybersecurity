# Network lab 1: Your first network (Network Starter)

Open Network Starter, cable four computers to a switch, give them addresses, and prove with ping which ones can talk and why one can't.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `netbuilder.json`, `README.txt`.

## 1. Open Network Starter and meet the two modes

Every network lab in this set happens in a network program in the Hacker Room, and you answer here as you go. Nothing to install. Labs 1 to 3 use **Network Starter**, a simple network builder for first lessons. In the Hacker Room open **Start → Workplace → Network lab** and pick lab 1: the switch, four computers and the lab's **Goals** load with it.

Network Starter has two modes. **Design** (the hammer) is where you place devices, cables and settings. **Simulation** (the play arrow) switches the network on, so you can sit at the computers and type commands.

- [ ] Network Starter open, with sw1 and four computers on the canvas
- [ ] Find Design and Simulation in the bar
- [ ] Find Computer, Switch and Cable in the bar, and the Goals tab

> Hint: You can't change an address in Simulation mode, and you can't ping in Design mode. Most beginner mistakes come from being in the wrong mode.

## 2. Cable four computers to a switch

In **Design** mode, choose **Cable**, click a computer, then click **sw1**: Network Starter uses the first free port, so there's no counting ports. Cable all four, then press **Select**.

Click each computer and type its **IP address** and **Subnet mask** in the panel on the right, from the plan in README.txt. Give the fourth one, stray, the wrong address on purpose: 192.168.2.14.

## 3. Ping, and explain the one that fails

Press **Simulation** and click **till-1**: its desktop opens. Open **Command line** and type `ipconfig`, then `ping 192.168.1.12` and `ping 192.168.1.13`. Then `ping 192.168.2.14`.

The first two answer. The last one says **PING: transmit failed. General failure.**, even though the cable and the switch are fine: 192.168.2.14 is in a different network (192.168.2.0/24), and till-1 has no gateway to send it to.

- [ ] ipconfig shows till-1's address and subnet mask
- [ ] Pings to till-2 and office get replies
- [ ] Ping to stray fails: write one line on why

> Hint: Same cable, same switch, different network number. Computers only talk directly to addresses inside their own network.

## 4. Fix it and watch the data exchange

Press **Design**, click stray and change its address to 192.168.1.14. Press **Simulation**, click till-1 and ping it: it answers now.

Open the **Data exchange** tab for till-1, press **Clear**, ping stray once more, and read the list from the top. Before the first ping you'll see **ARP**: till-1 asks "who has 192.168.1.14?" and stray answers with its MAC address. Then come the ICMP echo requests and echo replies. Click any row to see its layers.

- [ ] stray moved to 192.168.1.14 and it answers
- [ ] Data exchange open for till-1
- [ ] Find the ARP question and answer before the first ICMP packet

> Hint: ARP happens once, before the first packet: the answer is remembered for a while, so the next pings go straight out.
