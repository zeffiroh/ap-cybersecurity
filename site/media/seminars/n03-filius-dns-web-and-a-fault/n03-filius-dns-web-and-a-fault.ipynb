# Network lab 3: DNS, a web server and a fault to find (Filius)

Add a web server and a DNS server, open the shop's site by name, then troubleshoot a computer that can't, one layer at a time.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `README.txt`.

## 1. Add a web server and a DNS server

Open your lab 2 network and save it under a new name. Add two computers to the back-office switch: **web01** (192.168.2.80) and **dns01** (192.168.2.53), gateway 192.168.2.1.

In simulation mode, install **Web server** on web01 and start it. Install **DNS server** on dns01, add a record **www.banyan.lan → 192.168.2.80** (type A), and start it.

- [ ] web01 and dns01 added with gateways
- [ ] Web server running on web01
- [ ] DNS record www.banyan.lan → 192.168.2.80

> Hint: A DNS "A record" maps a name to an IPv4 address.

?> Which protocol turns www.banyan.lan into an address? (three letters) :: sha256:dd75a9d6fb309c4399fe425cd5f90ff95eba135d6924fb91766ee5d3726b168a

## 2. Open the site by name

Set every computer's **DNS server** to 192.168.2.53. On till-1, install **Web browser** and open `http://www.banyan.lan`. Then open **Show data exchange** on till-1 and find the DNS question before the web request.

- [ ] Every computer uses 192.168.2.53 for DNS
- [ ] till-1 opens the site by name
- [ ] DNS query found before the HTTP request

> Hint: If the name fails but `http://192.168.2.80` works, the problem is DNS, not the network.

?> Which address does the DNS server return for www.banyan.lan? :: sha256:57cd9fa0ab5ac30d60ffe69dd598a443421973c0765e5541b35fa2b4866ab774

## 3. Find the fault

Go to design mode and break till-2 in one way: change its gateway to **192.168.1.100**. Now pretend you don't know. On till-2, work up the layers and write what each test shows:

1. `ipconfig`: is the address right?
2. `ping 192.168.1.12` (itself) and `ping 192.168.1.11` (a neighbour): is the local network fine?
3. `ping 192.168.1.1`: is the gateway right?
4. `ping 192.168.2.53`, then the site by name.

Fix it and test again.

- [ ] Each test written down with its result
- [ ] The failing layer named
- [ ] Fixed and tested again

> Hint: Work from the bottom up. If the neighbour answers but the other network doesn't, the cable and address are fine: look at the gateway.
