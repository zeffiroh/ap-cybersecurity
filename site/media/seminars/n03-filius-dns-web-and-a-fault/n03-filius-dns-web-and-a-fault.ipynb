# Network lab 3: DNS, a web server and a fault (Network Starter)

Add a web server and a DNS server, open the shop's site by name, then troubleshoot a computer that can't, one layer at a time.

The files for this seminar are in `~/case` (Linux) and `case/` (Python): `netbuilder.json`, `README.txt`.

## 1. Set up a web server and a DNS server

Your lab 2 network is already built, and two servers are cabled to the back-office switch. In **Design** mode, click **web01** and give it 192.168.2.80 with gateway 192.168.2.1, then **dns01**: 192.168.2.53, gateway 192.168.2.1.

Press **Simulation**. Click web01, open **Web server** on its desktop and press **Start**. Click dns01, open **DNS server**, add the record **www.banyan.lan → 192.168.2.80** and press **Add**, then **Start**.

- [ ] web01 and dns01 addressed with gateways
- [ ] Web server running on web01
- [ ] DNS record www.banyan.lan → 192.168.2.80, DNS server running

> Hint: A record that maps a name to an IPv4 address is called an "A record".

## 2. Open the site by name

Press **Design** and set every computer's **DNS server** to 192.168.2.53, the servers too. Press **Simulation**, click till-1, open **Web browser** and go to `http://www.banyan.lan`. Then open the **Data exchange** tab for till-1 and find the DNS question before the web request.

- [ ] Every computer uses 192.168.2.53 for DNS
- [ ] till-1 opens the site by name
- [ ] DNS query found before the HTTP request

> Hint: If the name fails but `http://192.168.2.80` works, the problem is DNS, not the network.

?> Which address does the DNS server return for www.banyan.lan? :: sha256:57cd9fa0ab5ac30d60ffe69dd598a443421973c0765e5541b35fa2b4866ab774

## 3. Find the fault

Press **Design** and break till-2 in one way: change its gateway to **192.168.1.100**. Now pretend you don't know. Press **Simulation**, open till-2's **Command line**, work up the layers and write what each test shows:

1. `ipconfig`: is the address right?
2. `ping 192.168.1.12` (itself) and `ping 192.168.1.11` (a neighbour): is the local network fine?
3. `ping 192.168.1.1`: is the router there? Compare it with the gateway `ipconfig` showed.
4. `ping 192.168.2.53`, then `nslookup www.banyan.lan`, then the site in the **Web browser**.

Fix it and test again.

- [ ] Each test written down with its result
- [ ] The failing layer named
- [ ] Fixed and tested again

> Hint: Work from the bottom up. If the neighbour answers but the other network doesn't, the cable and address are fine: look at the gateway.
